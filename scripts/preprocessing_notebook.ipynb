A través del uso de Python en Google Colab, se busca transformar datos en información útil que apoye la toma de decisiones académicas.

```python
import pandas as pd
import numpy as np

# Cargamos el dataset usando el encoding correcto ya que contiene caracteres especiales o en español
main_df = pd.read_csv('../dataset/raw/dataset_unicaribe.csv', encoding="latin-1")

# Observamos la forma del DataFrame, los tipos de datos y las primeras filas para entender mejor el contenido del dataset
print(main_df.shape)
print("-------------")
print(main_df.dtypes)
print("-------------")
main_df.head(10)
```

### Proceso de Limpieza de Datos

```python
# ------------------------------
# 1. Normalizando nombres de columnas
# ------------------------------

# Renombrando explícitamente para consistencia

main_df = main_df.rename(columns={
    "Edad": "edad",
    "Género": "genero",
    "Carrera": "carrera",
    "Materias Inscritas": "materias_inscritas",
    "Horas de Estudio por Semana": "horas_estudio_semana",
    "Asistencia (%)": "asistencia",
    "Uso de Plataforma Virtual": "uso_plataformas",
    "Promedio (GPA)": "gpa",
    "Trabajo": "trabajo",
    "Acceso a Internet": "internet"
})
# ------------------------------
# 2. Limpieza de valores categóricos
# ------------------------------

# Género
main_df["genero"] = (
    main_df["genero"]
    .replace({"Masculino": "M", "Femenino": "F"})
    .infer_objects(copy=False)
)

# Trabajo
main_df["trabajo"] = (
    main_df["trabajo"]
    .replace({"Sí": 1, "No": 0})
    .infer_objects(copy=False)
)

# Internet
main_df["internet"] = (
    main_df["internet"]
    .replace({"Sí": 1, "No": 0})
    .infer_objects(copy=False)
)

# ------------------------------
# 3. Limpieza de valores numéricos
# ------------------------------
# Convertir a numérico
numeric_cols = ["edad", "materias_inscritas", "horas_estudio_semana", "asistencia", "gpa"]

for col in numeric_cols:
    main_df[col] = pd.to_numeric(main_df[col], errors="coerce")

# ------------------------------
# 4. Eliminación de duplicados
# ------------------------------
main_df.drop_duplicates(inplace=True)


# ------------------------------
# 5. Validación final de datos
# ------------------------------
print("\nResumen del dataset limpio:")
print(main_df.info())
print("-------------")

print("\nValores nulos:")
print(main_df.isnull().sum())
print("-------------")

print("\nPrimeras filas:")
print(main_df.shape)
print("-------------")
print(main_df.dtypes)
print("-------------")
main_df.head(10)

# ------------------------------
# 6. Exportamos el dataset limpio a un nuevo archivo CSV
# ------------------------------
main_df.to_csv("../dataset/cleaned/cl_dataset_unicaribe.csv", index=False, encoding="latin-1")
print("\nDataset limpio guardado como 'cl_dataset_unicaribe.csv'")
```
